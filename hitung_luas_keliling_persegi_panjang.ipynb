```python
%%writefile luaskelilingpersegipanjang.cpp
#include <iostream>
using namespace std;

int main() {
    int P, L, Ls, Lk;

    cout << "Masukkan P: ";
    cin >> P;
    cout << "Masukkan L: ";
    cin >> L;

    Ls = P * L;
    Lk = 2 * (P + L);

    cout << "Hasil luas persegi panjang: " << Ls << endl;
    cout << "Hasil keliling persegi panjang: " << Lk << endl;
    return 0;
}
```
